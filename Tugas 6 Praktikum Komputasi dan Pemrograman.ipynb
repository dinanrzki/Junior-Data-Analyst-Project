```python
# Metode Jacobi
# Soal Nomor 2

# Tebakan awal
x1 = 0.0
x2 = 0.0
x3 = 0.0

print("Iterasi 0:")
print(f"x1 = {x1:.6f}, x2 = {x2:.6f}, x3 = {x3:.6f}")

for k in range(1, 6):

    x1_baru = (16 - 2*x2 - x3)/5
    x2_baru = (7 - 5*x1 + 2*x3)/6
    x3_baru = (16 + x1 + x2)/4

    print(f"\nIterasi {k}:")
    print(f"x1 = {x1_baru:.6f}")
    print(f"x2 = {x2_baru:.6f}")
    print(f"x3 = {x3_baru:.6f}")

    x1 = x1_baru
    x2 = x2_baru
    x3 = x3_baru

print("\nHasil Iterasi ke-5")
print(f"x1 = {x1:.6f}")
print(f"x2 = {x2:.6f}")
print(f"x3 = {x3:.6f}")
```
