```python
import numpy as np
def court_lu(A) :
  n = len(A)
  L = np.zeros( (n, n))
  U = np.eye(n)
  for j in range (n) :
    for i in range (j, n):
      s = sum(L[i][k] * U [k][j] for k in range (j))
      L[i][j] = A [i][j] - s
    for i in range (j + 1, n):
      s = sum(L[j][k]* U[k][i] for k in range (j))
      U[j][i] = (A [j][i] - s) / L [j][j]
  return L, U
def forward_subtitutional(L, b):
  n = len(b)
  y = np.zeros (n)
  for i in range (n):
    y[i] = (b[i] - sum(L[i][j] * y[j] for j in range (i))) / L[i][i]
  return y
def backward_substitutional(U, y):
  n = len(y)
  x = np.zeros(n)
  for i in range (n - 1, -1, -1) :
    x[i] = y[i] - sum(U[i][j] * x[j] for j in range (i + 1, n))
  return x
A = np.array([
    [10, -2, -1, -1],
   [-2, 10, -1, -1],
  [-1, -1, 10, -2],
    [-1, -1, -2, 10]
], dtype = float)
b = np.array([3, 15, 27, -9], dtype = float)
L, U = court_lu(A)
y = forward_subtitutional(L, b)
x = backward_substitutional(U, y)
print("matriks L :")
print(L)
print("matriks U:")
print(U)
print("Vektor y:")
print(y)
for i in range(len(x)) :
  print(f"x[{i + 1}] = {x[i]: .6f}")
```
